```python
  #Convert date and datetime to datetme format
df['date']=pd.to_datetime(df['date'])
df['datetime']=pd.to_datetime(df['datetime'])


```

#**1. What is the distribution of coffee sales by type?**



```python
coffee_counts = df['coffee_name'].value_counts()

plt.figure(figsize=(10, 6))
coffee_counts.plot(kind='bar', color='skyblue')
plt.title('Distribution of Coffee Sales by Type')
plt.xlabel('Coffee Type')
plt.ylabel('Number of Sales')
plt.xticks(rotation=45)
plt.show()
```

With the analysis of above graph , we came to the conclusion that "Americano with Milk" has the highest sales

#**2. What is the average money spent for each coffee type?**


```python
avg_spent = df.groupby('coffee_name')['money'].mean()

plt.figure(figsize=(10, 6))
avg_spent.plot(kind='bar', color='green')
plt.title('Average Money Spent by Coffee Type')
plt.xlabel('Coffee Type')
plt.ylabel('Average Money Spent')
plt.xticks(rotation=45)
plt.show()
```

# **3. How does payment method distribution look?**

```python
payment_counts = df['cash_type'].value_counts()

plt.figure(figsize=(6, 6))
payment_counts.plot(kind='pie', autopct='%1.1f%%', colors=['gold', 'lightblue'])
plt.title('Payment Method Distribution')
plt.ylabel('')
plt.show()
```

Most payments are done using card

# **4. How do total sales compare across different dates?**

```python
date_sales = df.groupby('date')['money'].sum()


plt.figure(figsize=(16, 8))
date_sales.plot(kind='line', marker='o', color='purple')
plt.title('Total Sales Across Different Dates')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

# **5. What are the top 5 most purchased coffees?**

```python
top_5_coffees = coffee_counts.head(5)

plt.figure(figsize=(8, 6))
top_5_coffees.plot(kind='barh', color='orange')
plt.title('Top 5 Most Purchased Coffees')
plt.xlabel('Number of Sales')
plt.ylabel('Coffee Type')
plt.gca().invert_yaxis()
plt.show()
```

# **6. What is the total revenue generated by each coffee type?**

```python
revenue = df.groupby(['coffee_name']).sum(['money']).sort_values(by='money', ascending=False)
revenue
```

```python
plt.figure(figsize=(10,4))
ax = sns.barplot(data=revenue, x='money', y='coffee_name', color='steelblue')
ax.bar_label(ax.containers[0], fontsize=6)
plt.xlabel("Revenue")
```

The product with the highest revenue is Americano with Milk whereas Espresso is at the bottom

# **7. How many sales were made for cash vs. card?**

```python
payment_counts = df.groupby('cash_type')['money'].count()

plt.figure(figsize=(6, 6))
payment_counts.plot(kind='bar', color=['blue', 'gray'])
plt.title('Sales by Payment Method (Cash vs. Card)')
plt.xlabel('Payment Method')
plt.ylabel('Number of Sales')
plt.xticks(rotation=0)
plt.show()
```

Payments are done using card with more than 1000 Transactions

# **8. What are the monthly sales for each coffee?**

```python
monthly_sales = df.groupby(['coffee_name','month']).count()['date'].reset_index().rename(columns={'date':'count'}).pivot(index='month', columns='coffee_name',values='count').reset_index()
monthly_sales
```

```python
monthly_sales.describe().T.loc[:,['min','max']]
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(data=monthly_sales)
plt.legend(loc='upper left')
plt.title("Monthly Sales")
plt.xticks(range(len(monthly_sales['month'])),monthly_sales['month'],size='small')
plt.show()
```

As shown in the line chart above, Americano with Milk and Latte, and Cappuccino are top selling coffee
 types, while Cocoa and Expresso have lowest sales. Additionally, Americano with Milk and Latte show an
 upward trending.

# **9. What are the weekday sales for coffee?**

```python
weekday_sales = df.groupby(['day']).count()['date'].reset_index().rename(columns={'date':'count'})
weekday_sales
```

```python
plt.figure(figsize=(12,6))
sns.barplot(data=weekday_sales, x='day' , y='count', color='steelblue')
plt.legend(loc='upper left')
plt.title("Weekday Sales")
plt.xticks(range(len(weekday_sales['day'])),['Sun','Mon','Tue','Wed','Thur','Fri','Sat'],size='small')
plt.show()
```

 The bar chart reveals that Tuesday has the highest sales of the week, while sales on the other days are
 relatively similar.

# **10. What are the total sales by hour of the day?**

```python
df['hour'] = pd.to_datetime(df['datetime'], errors='coerce').dt.hour
hourly_sales = df.groupby('hour')['money'].sum()


plt.figure(figsize=(10, 6))
hourly_sales.plot(kind='line', marker='o', color='darkgreen')
plt.title('Total Sales by Hour of the Day')
plt.xlabel('Hour')
plt.ylabel('Total Sales')
plt.grid()
plt.show()
```

The plots above illustrate the shopping traffic for each product throughout the day. Notably, all products
 experience a peak in traffic around 10:00 AM, with this trend being particularly pronounced for Latte.
 Additionally, Cappuccino, Cocoa, and Hot Chocolate tend to be more popular during the evening hours,
 specifically between 6:00pm and 8:00pm.
